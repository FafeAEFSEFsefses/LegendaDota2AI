```python
# HW07.ipynb - Full Clustering Analysis for 3 Datasets
# ====================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.metrics import silhouette_score, davies_bouldin_score, calinski_harabasz_score
from sklearn.pipeline import Pipeline
import warnings
warnings.filterwarnings('ignore')
import json
import os
from itertools import cycle
import gc

# Устанавливаем стиль графиков
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")

# Создаем директории если их нет
os.makedirs('artifacts/figures', exist_ok=True)
os.makedirs('artifacts/labels', exist_ok=True)
os.makedirs('data', exist_ok=True)

# ============================================
# ФУНКЦИИ ДЛЯ ПОВТОРЯЮЩИХСЯ ЗАДАЧ
# ============================================

def load_and_preprocess(dataset_path, dataset_name):
    """Загрузка и предобработка данных"""
    print(f"\nЗагружаем {dataset_name}...")

    # Пробуем разные способы чтения файла
    try:
        # Пробуем прочитать как CSV
        df = pd.read_csv(dataset_path)
    except:
        try:
            # Пробуем с запятой как разделителем
            df = pd.read_csv(dataset_path, sep=',')
        except:
            try:
                # Пробуем с табуляцией
                df = pd.read_csv(dataset_path, sep='\t')
            except:
                # Пробуем с пробелом
                df = pd.read_csv(dataset_path, sep='\s+')

    print(f"Размер датасета: {df.shape}")
    print(f"Колонки: {df.columns.tolist()}")

    # Базовый анализ
    print("\nИнформация о датасете:")
    print(df.info())

    # Проверяем пропуски
    missing = df.isnull().sum()
    total_missing = missing.sum()
    if total_missing > 0:
        print(f"\nНайдено пропусков: {total_missing}")
        print("Столбцы с пропусками:")
        for col in missing[missing > 0].index:
            print(f"  {col}: {missing[col]} пропусков")
        # Заполняем медианой для числовых столбцов
        for col in df.select_dtypes(include=[np.number]).columns:
            if df[col].isnull().any():
                df[col].fillna(df[col].median(), inplace=True)
        print("Пропуски заполнены медианами")
    else:
        print("\nПропусков не найдено")

    # Отделяем sample_id и признаки
    if 'sample_id' in df.columns:
        print("Найден столбец sample_id")
        sample_ids = df['sample_id']
        X = df.drop('sample_id', axis=1)
    else:
        print("Столбец sample_id не найден, создаем искусственные ID")
        sample_ids = pd.Series(range(len(df)))
        X = df.copy()

    print(f"Признаков: {X.shape[1]}")

    # Масштабирование
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)

    return X_scaled, X.columns.tolist(), sample_ids, df

def run_kmeans_analysis(X_scaled, k_range=range(2, 11), dataset_name=""):
    """Анализ KMeans с подбором k"""
    print(f"\n{'='*60}")
    print(f"KMEANS АНАЛИЗ - {dataset_name}")
    print(f"{'='*60}")

    silhouette_scores = []
    db_scores = []
    ch_scores = []

    for k in k_range:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        labels = kmeans.fit_predict(X_scaled)

        if len(np.unique(labels)) > 1:
            silhouette_scores.append(silhouette_score(X_scaled, labels))
            db_scores.append(davies_bouldin_score(X_scaled, labels))
            ch_scores.append(calinski_harabasz_score(X_scaled, labels))
        else:
            silhouette_scores.append(np.nan)
            db_scores.append(np.nan)
            ch_scores.append(np.nan)

    # Находим лучшее k
    valid_scores = [s for s in silhouette_scores if not np.isnan(s)]
    if valid_scores:
        best_k = k_range[np.nanargmax(silhouette_scores)]
    else:
        best_k = 2

    # Обучаем с лучшим k
    kmeans_best = KMeans(n_clusters=best_k, random_state=42, n_init=10)
    kmeans_labels = kmeans_best.fit_predict(X_scaled)

    # Собираем результаты
    results = {
        'best_k': best_k,
        'labels': kmeans_labels,
        'model': kmeans_best,
        'metrics': {
            'silhouette': silhouette_score(X_scaled, kmeans_labels) if len(np.unique(kmeans_labels)) > 1 else 0,
            'davies_bouldin': davies_bouldin_score(X_scaled, kmeans_labels) if len(np.unique(kmeans_labels)) > 1 else 0,
            'calinski_harabasz': calinski_harabasz_score(X_scaled, kmeans_labels) if len(np.unique(kmeans_labels)) > 1 else 0,
        },
        'all_scores': {
            'k_range': list(k_range),
            'silhouette': silhouette_scores,
            'davies_bouldin': db_scores,
            'calinski_harabasz': ch_scores,
        }
    }

    print(f"Лучшее k: {best_k}")
    print(f"Silhouette Score: {results['metrics']['silhouette']:.4f}")
    print(f"Davies-Bouldin Score: {results['metrics']['davies_bouldin']:.4f}")
    print(f"Calinski-Harabasz Score: {results['metrics']['calinski_harabasz']:.4f}")

    return results

def run_dbscan_analysis(X_scaled, eps_range=np.arange(0.1, 3.0, 0.1),
                        min_samples_range=[2, 3, 5, 7, 10], dataset_name=""):
    """Анализ DBSCAN с подбором параметров"""
    print(f"\n{'='*60}")
    print(f"DBSCAN АНАЛИЗ - {dataset_name}")
    print(f"{'='*60}")

    best_score = -1
    best_params = None
    best_labels = None
    best_model = None

    results_list = []

    for eps in eps_range:
        for min_samples in min_samples_range:
            dbscan = DBSCAN(eps=eps, min_samples=min_samples)
            labels = dbscan.fit_predict(X_scaled)

            # Исключаем шумные точки для расчета метрик
            non_noise_mask = labels != -1
            n_clusters = len(np.unique(labels[non_noise_mask]))

            if n_clusters > 1 and np.sum(non_noise_mask) > 1:
                silhouette = silhouette_score(X_scaled[non_noise_mask], labels[non_noise_mask])
                noise_ratio = np.sum(labels == -1) / len(labels)

                results_list.append({
                    'eps': eps,
                    'min_samples': min_samples,
                    'silhouette': silhouette,
                    'noise_ratio': noise_ratio,
                    'n_clusters': n_clusters,
                    'labels': labels,
                    'model': dbscan
                })

                if silhouette > best_score:
                    best_score = silhouette
                    best_params = {'eps': eps, 'min_samples': min_samples}
                    best_labels = labels
                    best_model = dbscan

    if best_params:
        # Рассчитываем все метрики для лучшей модели
        non_noise_mask = best_labels != -1
        if np.sum(non_noise_mask) > 1 and len(np.unique(best_labels[non_noise_mask])) > 1:
            silhouette = silhouette_score(X_scaled[non_noise_mask], best_labels[non_noise_mask])
            db = davies_bouldin_score(X_scaled[non_noise_mask], best_labels[non_noise_mask])
            ch = calinski_harabasz_score(X_scaled[non_noise_mask], best_labels[non_noise_mask])
        else:
            silhouette, db, ch = 0, 0, 0

        results = {
            'best_params': best_params,
            'labels': best_labels,
            'model': best_model,
            'metrics': {
                'silhouette': silhouette,
                'davies_bouldin': db,
                'calinski_harabasz': ch,
                'noise_ratio': np.sum(best_labels == -1) / len(best_labels),
                'n_clusters': len(np.unique(best_labels[best_labels != -1])),
            },
            'all_results': results_list
        }

        print(f"Лучшие параметры: eps={best_params['eps']:.1f}, min_samples={best_params['min_samples']}")
        print(f"Silhouette Score: {silhouette:.4f}")
        print(f"Доля шума: {results['metrics']['noise_ratio']:.2%}")
        print(f"Количество кластеров: {results['metrics']['n_clusters']}")

        return results
    else:
        print("Не удалось найти хорошие параметры для DBSCAN")
        return None

def run_agglomerative_analysis(X_scaled, k_range=range(2, 11),
                               linkage_methods=['ward', 'complete', 'average', 'single'],
                               dataset_name=""):
    """Анализ Agglomerative Clustering"""
    print(f"\n{'='*60}")
    print(f"AGGLOMERATIVE АНАЛИЗ - {dataset_name}")
    print(f"{'='*60}")

    best_score = -1
    best_params = None
    best_labels = None
    best_model = None

    for linkage in linkage_methods:
        linkage_scores = []

        for k in k_range:
            agg = AgglomerativeClustering(n_clusters=k, linkage=linkage)
            labels = agg.fit_predict(X_scaled)

            if len(np.unique(labels)) > 1:
                silhouette = silhouette_score(X_scaled, labels)
                linkage_scores.append(silhouette)
            else:
                linkage_scores.append(-1)

        # Находим лучшее k для этого linkage
        if linkage_scores:
            best_k_idx = np.argmax(linkage_scores)
            best_k = k_range[best_k_idx]
            best_silhouette = linkage_scores[best_k_idx]

            if best_silhouette > best_score:
                best_score = best_silhouette
                best_params = {'linkage': linkage, 'k': best_k}

                # Обучаем с лучшими параметрами
                best_model = AgglomerativeClustering(n_clusters=best_k, linkage=linkage)
                best_labels = best_model.fit_predict(X_scaled)

    if best_params:
        # Рассчитываем все метрики
        silhouette = silhouette_score(X_scaled, best_labels) if len(np.unique(best_labels)) > 1 else 0
        db = davies_bouldin_score(X_scaled, best_labels) if len(np.unique(best_labels)) > 1 else 0
        ch = calinski_harabasz_score(X_scaled, best_labels) if len(np.unique(best_labels)) > 1 else 0

        results = {
            'best_params': best_params,
            'labels': best_labels,
            'model': best_model,
            'metrics': {
                'silhouette': silhouette,
                'davies_bouldin': db,
                'calinski_harabasz': ch,
                'n_clusters': best_params['k'],
            }
        }

        print(f"Лучшая конфигурация: linkage={best_params['linkage']}, k={best_params['k']}")
        print(f"Silhouette Score: {silhouette:.4f}")
        print(f"Davies-Bouldin Score: {db:.4f}")
        print(f"Calinski-Harabasz Score: {ch:.4f}")

        return results
    else:
        print("Не удалось найти хорошие параметры для Agglomerative Clustering")
        return None

def plot_clustering_results(X_pca, labels_dict, titles, save_path):
    """Визуализация результатов кластеризации"""
    fig, axes = plt.subplots(2, 2, figsize=(15, 12))
    axes = axes.flatten()

    colors = cycle(['tab:blue', 'tab:orange', 'tab:green', 'tab:red', 'tab:purple',
                    'tab:brown', 'tab:pink', 'tab:gray', 'tab:olive', 'tab:cyan'])

    for idx, (title, labels) in enumerate(zip(titles, labels_dict)):
        ax = axes[idx]

        if title.startswith('DBSCAN'):
            noise_mask = labels == -1
            cluster_mask = ~noise_mask

            if np.any(noise_mask):
                ax.scatter(X_pca[noise_mask, 0], X_pca[noise_mask, 1],
                          c='gray', alpha=0.3, s=10, label='Noise')

            unique_labels = np.unique(labels[cluster_mask])
            for cluster_label in unique_labels:
                if cluster_label != -1:
                    cluster_points = labels == cluster_label
                    color = next(colors)
                    ax.scatter(X_pca[cluster_points, 0], X_pca[cluster_points, 1],
                              c=[color], alpha=0.6, s=30, label=f'Cluster {cluster_label}')
        else:
            scatter = ax.scatter(X_pca[:, 0], X_pca[:, 1], c=labels,
                                cmap='tab10', alpha=0.6, s=30)
            plt.colorbar(scatter, ax=ax)

        ax.set_title(title, fontsize=12)
        ax.set_xlabel('PC1')
        ax.set_ylabel('PC2')
        ax.grid(True, alpha=0.3)

        if title.startswith('DBSCAN') and len(np.unique(labels[labels != -1])) <= 5:
            ax.legend(loc='upper right', fontsize=8)

    plt.tight_layout()
    plt.savefig(save_path, dpi=300, bbox_inches='tight')
    plt.show()
    plt.close()

# ============================================
# ОСНОВНОЙ ЦИКЛ ОБРАБОТКИ ДАТАСЕТОВ
# ============================================

# Создаем тестовые данные для примера
print("Создание тестовых данных...")
# Создаем папку data если её нет
os.makedirs('data', exist_ok=True)

# Для тестирования создадим 3 простых CSV файла если их нет
import sys

# Проверяем, есть ли файлы в папке data
existing_files = []
for i in range(1, 4):
    csv_file = f'data/S07-hw-dataset-0{i}.csv'
    txt_file = f'data/S07-hw-dataset-0{i}.txt'

    if os.path.exists(csv_file):
        existing_files.append(csv_file)
        print(f"Найден файл: {csv_file}")
    elif os.path.exists(txt_file):
        existing_files.append(txt_file)
        print(f"Найден файл: {txt_file}")
    else:
        print(f"Файл dataset-0{i} не найден, создаем тестовый...")
        # Создаем тестовые данные
        np.random.seed(42 + i)
        n_samples = 200

        if i == 1:
            # Простые сферические кластеры
            X1 = np.random.randn(n_samples, 2) * 0.5 + np.array([0, 0])
            X2 = np.random.randn(n_samples, 2) * 0.5 + np.array([3, 3])
            X3 = np.random.randn(n_samples, 2) * 0.5 + np.array([-3, 3])
            X = np.vstack([X1, X2, X3])
        elif i == 2:
            # Нелинейные кластеры (полумесяцы)
            from sklearn.datasets import make_moons, make_circles
            X1, _ = make_moons(n_samples=n_samples, noise=0.05)
            X2, _ = make_circles(n_samples=n_samples, factor=0.5, noise=0.05)
            X2 = X2 * 2 + np.array([1.5, 0])
            X = np.vstack([X1, X2])
        else:
            # Кластеры разной плотности
            X1 = np.random.randn(n_samples//2, 2) * 0.3 + np.array([0, 0])
            X2 = np.random.randn(n_samples//2, 2) * 1.0 + np.array([3, 3])
            X3 = np.random.randn(n_samples, 2) * 0.1 + np.array([-3, 3])
            X = np.vstack([X1, X2, X3])

        # Добавляем шум
        n_noise = int(0.1 * len(X))
        noise = np.random.randn(n_noise, 2) * 2
        X = np.vstack([X, noise])

        # Создаем DataFrame
        df = pd.DataFrame(X, columns=[f'f{j:02d}' for j in range(X.shape[1])])
        df.insert(0, 'sample_id', range(len(df)))

        # Сохраняем
        df.to_csv(csv_file, index=False)
        existing_files.append(csv_file)
        print(f"Создан тестовый файл: {csv_file}")

# Используем найденные или созданные файлы
datasets = []
for i, file_path in enumerate(existing_files, 1):
    datasets.append({
        'name': f'dataset_0{i}',
        'path': file_path,
        'description': f'Synthetic dataset {i} for testing'
    })

# Словарь для хранения всех результатов
all_results = {}
best_configs = {}
metrics_summary = {}

# Обрабатываем каждый датасет
for dataset_info in datasets:
    dataset_name = dataset_info['name']
    dataset_path = dataset_info['path']

    print(f"\n{'='*80}")
    print(f"НАЧИНАЕМ АНАЛИЗ ДАТАСЕТА: {dataset_name.upper()}")
    print(f"{'='*80}")

    # 1. Загрузка и предобработка данных
    X_scaled, feature_names, sample_ids, df = load_and_preprocess(dataset_path, dataset_name)

    # Сохраняем базовую информацию
    all_results[dataset_name] = {
        'X_scaled': X_scaled,
        'feature_names': feature_names,
        'sample_ids': sample_ids,
        'original_data': df
    }

    # 2. Выполняем анализ для каждого метода
    print(f"\n{'='*60}")
    print(f"АНАЛИЗ КЛАСТЕРИЗАЦИИ - {dataset_name}")
    print(f"{'='*60}")

    # KMeans
    kmeans_results = run_kmeans_analysis(X_scaled, dataset_name=dataset_name)
    all_results[dataset_name]['kmeans'] = kmeans_results

    # DBSCAN
    if '02' in dataset_name:
        eps_range = np.arange(0.1, 2.0, 0.1)
    elif '03' in dataset_name:
        eps_range = np.arange(0.1, 1.5, 0.1)
    else:
        eps_range = np.arange(0.1, 1.0, 0.1)

    dbscan_results = run_dbscan_analysis(X_scaled, eps_range=eps_range, dataset_name=dataset_name)
    all_results[dataset_name]['dbscan'] = dbscan_results

    # Agglomerative Clustering
    agglo_results = run_agglomerative_analysis(X_scaled, dataset_name=dataset_name)
    all_results[dataset_name]['agglomerative'] = agglo_results

    # 3. Сравниваем методы и выбираем лучший
    print(f"\n{'='*60}")
    print(f"СРАВНЕНИЕ МЕТОДОВ - {dataset_name}")
    print(f"{'='*60}")

    # Собираем метрики для сравнения
    metrics_data = {}

    if kmeans_results:
        metrics_data['KMeans'] = kmeans_results['metrics']

    if dbscan_results:
        metrics_data['DBSCAN'] = dbscan_results['metrics']

    if agglo_results:
        metrics_data['Agglomerative'] = agglo_results['metrics']

    # Находим лучший метод по Silhouette Score
    if metrics_data:
        best_method = max(metrics_data.keys(), key=lambda x: metrics_data[x]['silhouette'])
        best_metrics = metrics_data[best_method]

        print(f"\nЛучший метод: {best_method}")
        print(f"Silhouette Score: {best_metrics['silhouette']:.4f}")

        # Сохраняем лучшую конфигурацию
        if best_method == 'KMeans':
            best_params = f"k={kmeans_results['best_k']}"
            best_labels = kmeans_results['labels']
        elif best_method == 'DBSCAN':
            best_params = f"eps={dbscan_results['best_params']['eps']:.1f}, min_samples={dbscan_results['best_params']['min_samples']}"
            best_labels = dbscan_results['labels']
        else:
            best_params = f"linkage={agglo_results['best_params']['linkage']}, k={agglo_results['best_params']['k']}"
            best_labels = agglo_results['labels']

        best_configs[dataset_name] = {
            'best_method': best_method,
            'best_params': best_params,
            'best_metrics': best_metrics,
            'n_clusters': best_metrics.get('n_clusters', len(np.unique(best_labels[best_labels != -1])) if -1 in best_labels else len(np.unique(best_labels)))
        }

        # Сохраняем метки кластеров
        labels_df = pd.DataFrame({
            'sample_id': sample_ids,
            'cluster_label': best_labels
        })
        labels_df.to_csv(f'artifacts/labels/labels_hw07_{dataset_name}.csv', index=False)

        metrics_summary[dataset_name] = metrics_data

    # 4. Визуализация результатов
    print(f"\n{'='*60}")
    print(f"ВИЗУАЛИЗАЦИЯ - {dataset_name}")
    print(f"{'='*60}")

    # PCA для визуализации
    pca = PCA(n_components=2, random_state=42)
    X_pca = pca.fit_transform(X_scaled)
    print(f"Объясненная дисперсия PCA: {pca.explained_variance_ratio_.sum():.2%}")

    # Подготовка данных для визуализации
    titles = ['Исходные данные', 'KMeans', 'DBSCAN', 'Agglomerative']
    labels_dict = [
        np.zeros(len(X_scaled)),  # Все точки одного цвета
        kmeans_results['labels'] if kmeans_results else np.zeros(len(X_scaled)),
        dbscan_results['labels'] if dbscan_results else np.zeros(len(X_scaled)),
        agglo_results['labels'] if agglo_results else np.zeros(len(X_scaled))
    ]

    # Визуализируем результаты кластеризации
    plot_clustering_results(X_pca, labels_dict, titles,
                           f'artifacts/figures/{dataset_name}_clustering_results.png')

    # График для KMeans - Silhouette vs k
    if kmeans_results:
        plt.figure(figsize=(10, 6))
        plt.plot(kmeans_results['all_scores']['k_range'],
                kmeans_results['all_scores']['silhouette'], 'bo-', linewidth=2, markersize=8)
        plt.axvline(x=kmeans_results['best_k'], color='r', linestyle='--',
                   label=f'Best k = {kmeans_results["best_k"]}')
        plt.xlabel('Number of clusters (k)')
        plt.ylabel('Silhouette Score')
        plt.title(f'KMeans: Silhouette Score vs k - {dataset_name}')
        plt.grid(True, alpha=0.3)
        plt.legend()
        plt.savefig(f'artifacts/figures/{dataset_name}_kmeans_silhouette.png', dpi=300, bbox_inches='tight')
        plt.show()
        plt.close()

    print(f"\nАнализ датасета {dataset_name} завершен!")
    print(f"{'='*80}\n")

    # Очистка памяти
    gc.collect()

# ============================================
# ПРОВЕРКА УСТОЙЧИВОСТИ (для dataset_01)
# ============================================

print(f"\n{'='*80}")
print(f"ПРОВЕРКА УСТОЙЧИВОСТИ KMEANS ДЛЯ DATASET_01")
print(f"{'='*80}")

if 'dataset_01' in all_results:
    X_scaled_01 = all_results['dataset_01']['X_scaled']
    best_k_01 = all_results['dataset_01']['kmeans']['best_k']

    random_states = [0, 42, 123, 456, 789]
    silhouette_scores_stability = []
    db_scores_stability = []

    for rs in random_states:
        kmeans = KMeans(n_clusters=best_k_01, random_state=rs, n_init=10)
        labels = kmeans.fit_predict(X_scaled_01)

        if len(np.unique(labels)) > 1:
            silhouette_scores_stability.append(silhouette_score(X_scaled_01, labels))
            db_scores_stability.append(davies_bouldin_score(X_scaled_01, labels))

    # Анализ устойчивости
    if silhouette_scores_stability:
        silhouette_mean = np.mean(silhouette_scores_stability)
        silhouette_std = np.std(silhouette_scores_stability)
        silhouette_cv = silhouette_std / silhouette_mean if silhouette_mean != 0 else 0

        db_mean = np.mean(db_scores_stability)
        db_std = np.std(db_scores_stability)
        db_cv = db_std / db_mean if db_mean != 0 else 0

        print(f"\nУстойчивость KMeans (k={best_k_01}):")
        print(f"Silhouette Score:")
        print(f"  Среднее: {silhouette_mean:.4f}")
        print(f"  Стандартное отклонение: {silhouette_std:.4f}")
        print(f"  Коэффициент вариации: {silhouette_cv:.2%}")

        print(f"\nDavies-Bouldin Score:")
        print(f"  Среднее: {db_mean:.4f}")
        print(f"  Стандартное отклонение: {db_std:.4f}")
        print(f"  Коэффициент вариации: {db_cv:.2%}")

        # Визуализация устойчивости
        plt.figure(figsize=(12, 5))

        plt.subplot(1, 2, 1)
        plt.bar(range(len(random_states)), silhouette_scores_stability, color='blue', alpha=0.7)
        plt.axhline(y=silhouette_mean, color='r', linestyle='--', label=f'Mean: {silhouette_mean:.4f}')
        plt.xlabel('Random State')
        plt.ylabel('Silhouette Score')
        plt.title('Устойчивость Silhouette Score')
        plt.xticks(range(len(random_states)), random_states)
        plt.legend()
        plt.grid(True, alpha=0.3)

        plt.subplot(1, 2, 2)
        plt.bar(range(len(random_states)), db_scores_stability, color='red', alpha=0.7)
        plt.axhline(y=db_mean, color='b', linestyle='--', label=f'Mean: {db_mean:.4f}')
        plt.xlabel('Random State')
        plt.ylabel('Davies-Bouldin Score')
        plt.title('Устойчивость Davies-Bouldin Score')
        plt.xticks(range(len(random_states)), random_states)
        plt.legend()
        plt.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.savefig('artifacts/figures/dataset_01_stability_analysis.png', dpi=300, bbox_inches='tight')
        plt.show()
        plt.close()

# ============================================
# ИНТЕРПРЕТАЦИЯ КЛАСТЕРОВ (исправленная часть)
# ============================================

print(f"\n{'='*80}")
print(f"ИНТЕРПРЕТАЦИЯ КЛАСТЕРОВ")
print(f"{'='*80}")

for dataset_key in all_results.keys():
    X_scaled = all_results[dataset_key]['X_scaled']
    feature_names = all_results[dataset_key]['feature_names']

    # Используем лучший метод для интерпретации
    if dataset_key in best_configs:
        best_method = best_configs[dataset_key]['best_method']

        if best_method == 'KMeans':
            labels = all_results[dataset_key]['kmeans']['labels']
        elif best_method == 'DBSCAN':
            labels = all_results[dataset_key]['dbscan']['labels']
        else:
            labels = all_results[dataset_key]['agglomerative']['labels']

        # Создаем DataFrame для анализа
        df_analysis = pd.DataFrame(X_scaled, columns=feature_names)
        df_analysis['cluster'] = labels

        # Удаляем шумные точки из анализа (если есть)
        if -1 in labels:
            df_analysis = df_analysis[df_analysis['cluster'] != -1]

        # Анализ средних значений по кластерам (исправлено!)
        if not df_analysis.empty and 'cluster' in df_analysis.columns:
            # Используем groupby().mean() правильно
            cluster_means = df_analysis.groupby('cluster').mean()

            if not cluster_means.empty:
                print(f"\n{dataset_key.upper()} - Анализ средних значений по кластерам:")
                print(cluster_means.round(3))

                # Визуализация профилей кластеров (первые 4 признака)
                n_clusters = len(cluster_means)
                n_features = min(4, len(feature_names))

                fig, axes = plt.subplots(1, n_clusters, figsize=(4*n_clusters, 5))
                if n_clusters == 1:
                    axes = [axes]

                for i in range(n_clusters):
                    if n_clusters > 1:
                        ax = axes[i]
                    else:
                        ax = axes

                    cluster_data = cluster_means.iloc[i].values[:n_features]
                    feature_labels = feature_names[:n_features]

                    bars = ax.bar(range(n_features), cluster_data, alpha=0.7)
                    ax.axhline(y=0, color='k', linestyle='-', linewidth=0.5)
                    ax.set_title(f'Cluster {cluster_means.index[i]}')
                    ax.set_xticks(range(n_features))
                    ax.set_xticklabels(feature_labels, rotation=45)
                    ax.set_ylabel('Стандартизованное значение')
                    ax.grid(True, alpha=0.3)

                plt.suptitle(f'{dataset_key} - Профили кластеров (первые {n_features} признаков)', fontsize=14)
                plt.tight_layout()
                plt.savefig(f'artifacts/figures/{dataset_key}_cluster_profiles.png', dpi=300, bbox_inches='tight')
                plt.show()
                plt.close()

                # Heatmap средних значений
                plt.figure(figsize=(10, 6))
                sns.heatmap(cluster_means.T, annot=True, cmap='coolwarm', center=0,
                           fmt='.2f', cbar_kws={'label': 'Стандартизованное значение'})
                plt.title(f'{dataset_key} - Heatmap средних значений признаков по кластерам')
                plt.xlabel('Кластер')
                plt.ylabel('Признак')
                plt.tight_layout()
                plt.savefig(f'artifacts/figures/{dataset_key}_cluster_heatmap.png', dpi=300, bbox_inches='tight')
                plt.show()
                plt.close()

# ============================================
# СОХРАНЕНИЕ ИТОГОВЫХ АРТЕФАКТОВ
# ============================================

print(f"\n{'='*80}")
print(f"СОХРАНЕНИЕ ИТОГОВЫХ АРТЕФАКТОВ")
print(f"{'='*80}")

def convert_numpy(obj):
    """Преобразуем numpy типы в стандартные Python типы для JSON"""
    if isinstance(obj, np.integer):
        return int(obj)
    elif isinstance(obj, np.floating):
        return float(obj)
    elif isinstance(obj, np.ndarray):
        return obj.tolist()
    elif isinstance(obj, dict):
        return {k: convert_numpy(v) for k, v in obj.items()}
    elif isinstance(obj, list):
        return [convert_numpy(item) for item in obj]
    elif isinstance(obj, pd.Series):
        return obj.tolist()
    return obj

# 1. Сохраняем сводку метрик
if metrics_summary:
    with open('artifacts/metrics_summary.json', 'w') as f:
        json_metrics = convert_numpy(metrics_summary)
        json.dump(json_metrics, f, indent=2)
    print("✓ metrics_summary.json сохранен")

# 2. Сохраняем лучшие конфигурации
if best_configs:
    with open('artifacts/best_configs.json', 'w') as f:
        json_best_configs = convert_numpy(best_configs)
        json.dump(json_best_configs, f, indent=2)
    print("✓ best_configs.json сохранен")

# 3. Создаем сводный отчет
print("\nСВОДНЫЙ ОТЧЕТ:")
print("=" * 80)

for dataset_key in best_configs.keys():
    config = best_configs[dataset_key]
    print(f"\n{dataset_key.upper()}:")
    print(f"  Лучший метод: {config['best_method']}")
    print(f"  Параметры: {config['best_params']}")
    print(f"  Количество кластеров: {config['n_clusters']}")
    print(f"  Silhouette Score: {config['best_metrics']['silhouette']:.4f}")
    print(f"  Davies-Bouldin Score: {config['best_metrics']['davies_bouldin']:.4f}")
    if 'noise_ratio' in config['best_metrics']:
        print(f"  Доля шума: {config['best_metrics']['noise_ratio']:.2%}")

# 4. Подсчитываем созданные файлы
import glob
figures_count = len(glob.glob('artifacts/figures/*.png'))
labels_count = len(glob.glob('artifacts/labels/*.csv'))

print(f"\n{'='*80}")
print("АНАЛИЗ ЗАВЕРШЕН УСПЕШНО!")
print(f"{'='*80}")
print(f"\nСоздано файлов:")
print(f"  - Графиков: {figures_count} (минимум 6 требуется)")
print(f"  - Файлов с метками: {labels_count} (по одному на датасет)")
print(f"  - JSON файлов: 2 (metrics_summary.json, best_configs.json)")
print(f"\nВсе артефакты сохранены в папке 'artifacts/'")
```
